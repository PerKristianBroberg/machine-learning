```python
import numpy as np
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import joblib
import struct
```

```python
def load_images(filename):
    with open(filename, "rb") as f:
        magic, num, rows, cols = struct.unpack(">IIII", f.read(16))
        data = np.frombuffer(f.read(), dtype=np.uint8)
        images = data.reshape(num, rows * cols)
    return images

def load_labels(filename):
    with open(filename, "rb") as f:
        magic, num = struct.unpack(">II", f.read(8))
        labels = np.frombuffer(f.read(), dtype=np.uint8)
    return labels
```

```python

train_images = load_images("train-images-idx3-ubyte/train-images-idx3-ubyte")
train_labels = load_labels("train-labels-idx1-ubyte/train-labels-idx1-ubyte")
test_images  = load_images("t10k-images-idx3-ubyte/t10k-images-idx3-ubyte")
test_labels  = load_labels("t10k-labels-idx1-ubyte/t10k-labels-idx1-ubyte")

print("Data loaded:")
print(f"Train: {train_images.shape}, Test: {test_images.shape}")
```

```python
# Flatten (already flat, 784 features)
# Model was trained on black digits on white background, so keep 0–255 scale.
X_train = train_images.astype(np.float32)
y_train = train_labels
X_test = test_images.astype(np.float32)
y_test = test_labels
```

```python
# Normalize to 0–1
X_train = train_images / 255.0
X_test = test_images / 255.0

y_train, y_test = train_labels, test_labels
```

```python

print("Training improved MLP...")
model = MLPClassifier(
    hidden_layer_sizes=(300, 100),
    activation="relu",
    solver="adam",
    learning_rate_init=0.0005,
    alpha=1e-4,          # regularization
    batch_size=128,
    max_iter=40,         # train longer
    verbose=True,
    random_state=42
)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Test accuracy: {acc * 100:.2f}%")
```

```python
joblib.dump(model, "mnistmlp.pkl")
print("Model saved as mnistmlp.pkl")
```
